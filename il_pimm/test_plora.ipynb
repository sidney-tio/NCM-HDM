```python
from transformer import CTransformer
from plora import PLoraModel, PLoraConfig

transformer_configs = {
    "num_classes": 2,
    'emb_size': 256,
    "heads":4,
    "depth":4,
    "seq_length":256,
    "max_pool": False,
    "dropout": 0.1,
}


plora_configs = PLoraConfig(task_type="SEQ_CLS",
            inference_mode=False,
            r=8,
            lora_alpha=128,
            lora_dropout=0.1,
            num_virtual_users=256,
            user_token_dim=256,
            target_modules= ["toqueries", "tovalues"],
            bias= "none")

model = CTransformer(**transformer_configs)
```

```python
pconfigs = {'default': plora_configs}
model.config = transformer_configs
pmodel = PLoraModel(model, pconfigs, "default")
```
